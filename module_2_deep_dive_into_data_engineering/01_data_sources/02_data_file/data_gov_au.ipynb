Read Liquidator data

```python
# You code here
# 1. Read data from "LIQUIDATOR_202510" spreadsheet using Pandas
#    + Provide a schema when reading data file
# 2. Export to CSV as path 'destination/excel'

liquidator_schema = {
    "REGISTER_NAME": str,
    "LIQ_NUM": str,
    "OFF_LIQ_NUM": str, 
    "LIQ_NAME": str,
    "LIQ_START_DT": 'datetime64[ns]',
    "OFF_LIQ_START_DT": 'datetime64[ns]',
    "LIQ_STATUS": str,
    "LIQ_SUSP_DT": 'datetime64[ns]',
    "LIQ_ADD_LOCAL": str,
    "LIQ_ADD_STATE": str,
    "LIQ_ADD_PCODE": str,
    "LIQ_ADD_COUNTRY": str,
    "LIQ_FIRM": str
}

import os
import pandas as pd

date_cols = [col for col, dtype in liquidator_schema.items() if dtype == 'datetime64[ns]']
dtype_cols = {col: dtype for col, dtype in liquidator_schema.items() if dtype != 'datetime64[ns]'}

df_liquidator = pd.read_excel(
    excel_file_path,
    sheet_name=liquidator_sheetname,
    dtype=dtype_cols,          # enforce string columns
    parse_dates=date_cols,     # parse datetime columns
    # header=2                  # add this if your header starts on row 3 like earlier
)

print(df_liquidator.dtypes)    # quick check

# 3. Export to CSV under 'destination/excel'
output_folder = os.path.join("destination", "excel")
os.makedirs(output_folder, exist_ok=True)

output_file = os.path.join(output_folder, "liquidator_data.csv")
df_liquidator.to_csv(output_file, index=False)

print(f"Liquidator data exported to: {output_file}")
```
