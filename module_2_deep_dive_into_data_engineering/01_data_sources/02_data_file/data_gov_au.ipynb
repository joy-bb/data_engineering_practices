Data File - Excel

```python
# input variables
excel_file_path = "auditor_and_liquidator.xlsx"
auditor_sheetname = "REG_AUDITOR_202511"
liquidator_sheetname = "LIQUIDATOR_202510"

# output variables
path = "data_lake"
```

Read Auditor data

```python
# You code here
# 1. Read data from "REG_AUDITOR_202511" spreadsheet using Pandas
# 2. Apply data type conversion and export to CSV
# 3. Export to path 'destination/excel'

import os
import pandas as pd



df_auditor = pd.read_excel(excel_file_path, sheet_name=auditor_sheetname,header=2) #header = 2 do row 3, index se la 0,1,2 => header =2
df_auditor.dtypes
```

```python
# conver numeric identifier columns to string
# conver AFTER reading data => phuong phap nay khac voi pp sau day => can handle messy/real world data 

cols_to_str = ['REG_AUD_NUM', 'REG_AUD_ACN', 'REG_AUD_ADD_PCODE']

for col in cols_to_str:
    if col in df_auditor.columns:
        df_auditor[col] = (df_auditor[col].astype('Int64').astype('str'))

# conver date columns to date time format (dd/mm/yyyy)
cols_to_date = ['REG_AUD_START_DT', 'REG_AUD_SUSP_DT']

for col in cols_to_date:
    if col in df_auditor.columns:
        df_auditor[col] = pd.to_datetime(df_auditor[col], errors ='coerce', dayfirst= True)

df_auditor.dtypes
```

```python
import os

os.makedirs('destination/excel', exist_ok= True)

df_auditor.to_csv(f'destination/excel/{auditor_sheetname}.csv', index=False)
```

Read Liquidator data

```python
# You code here
# 1. Read data from "LIQUIDATOR_202510" spreadsheet using Pandas
#    + Provide a schema when reading data file
# 2. Export to CSV as path 'destination/excel'

# enforce schema first, define SCHEMA at READ time => faster, less memory, good for structure data

liquidator_schema = {
    "REGISTER_NAME": str,
    "LIQ_NUM": str,
    "OFF_LIQ_NUM": str, 
    "LIQ_NAME": str,
    "LIQ_START_DT": 'datetime64[ns]',
    "OFF_LIQ_START_DT": 'datetime64[ns]',
    "LIQ_STATUS": str,
    "LIQ_SUSP_DT": 'datetime64[ns]',
    "LIQ_ADD_LOCAL": str,
    "LIQ_ADD_STATE": str,
    "LIQ_ADD_PCODE": str,
    "LIQ_ADD_COUNTRY": str,
    "LIQ_FIRM": str
}

import os
import pandas as pd

date_cols = [col for col, dtype in liquidator_schema.items() if dtype == 'datetime64[ns]']
dtype_cols = {col: dtype for col, dtype in liquidator_schema.items() if dtype != 'datetime64[ns]'}

df_liquidator = pd.read_excel(
    excel_file_path,
    sheet_name=liquidator_sheetname,
    dtype=dtype_cols,          # enforce string columns
    parse_dates=date_cols,     # parse datetime columns
    # header=2                  # add this if your header starts on row 3 like earlier
)

print(df_liquidator.dtypes)    # quick check

# 3. Export to CSV under 'destination/excel'
output_folder = os.path.join("destination", "excel")
os.makedirs(output_folder, exist_ok=True)

output_file = os.path.join(output_folder, "liquidator_data.csv")
df_liquidator.to_csv(output_file, index=False)

print(f"Liquidator data exported to: {output_file}")
```
